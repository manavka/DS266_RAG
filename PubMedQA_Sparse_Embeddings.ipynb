```python
# load train and dev data

train = pd.read_json("/content/drive/MyDrive/DATASCI 266 - NLP/pqaa_train_set.json").T
dev = pd.read_json("/content/drive/MyDrive/DATASCI 266 - NLP/pqaa_dev_set.json").T
```

```python
train.head()
```

```python
contexts_train = [word.lower() for context in train['CONTEXTS'].tolist() for word in context] # list of documents for context

contexts_train_tokens = bm25s.tokenize(contexts_train, stopwords='en')

retriever = bm25s.BM25()
retriever.index(contexts_train_tokens)

query = "Is an advance care planning model feasible in community palliative care?"

query_tokens = bm25s.tokenize(query)

results, scores = retriever.retrieve(query_tokens, corpus=contexts_train, k=5)

for doc, score in zip(results[0], scores[0]):
    print(f"(Score: {score:.4f}) -> {doc}")
```
